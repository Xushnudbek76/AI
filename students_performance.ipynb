```python
df1 = pd.read_csv('/content/StudentsPerformance.csv')
```

```python
df1["total score"] = df1[["math score", "reading score", "writing score"]].sum(axis=1)
df1["average score"] = df1["total score"] / 3
df1["passed"] = np.where(df1["average score"] >= 50, "Yes", "No")

print("New columns preview (total, average, passed):")
print(df1[["total score", "average score", "passed"]].head())
```

```python
test_prep_group = df1.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()

print("student score by test preparation course:")
print(test_prep_group)



gender_avg = df1.groupby("gender")[["math score", "reading score", "writing score"]].mean()

print("student score by gender:")
print(gender_avg)

pass_rate = df1["passed"].value_counts(normalize=True) * 100

print("pass rate:")
print(pass_rate)
```

```python
scores = df1[["math score", "reading score", "writing score"]]


print("Numpy Mean:", np.mean(scores, axis=0))
print("Numpy Std:", np.std(scores, axis=0))

z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)

print("Z-score:")
print(z_score.head())
```

```python
top_scorers = df1[df1["average score"] >= 90];
print("Top scorers", top_scorers.head())

failed_scorers = df1[df1["passed"] == "No"];
print("Failed scorers", failed_scorers.head())
```
